```python
import pandas as pd
import numpy as np
print("Pandas imported successfully!")
```

```python
# Load the agricultural mandi dataset
df = pd.read_csv("../dataset/market_daily_sample.csv")
print("Dataset loaded successfully!")
```

```python
# Check dataset dimensions
print("Number of Rows:", df.shape[0]) # Gives the No of rows 
print("Number of Columns:", df.shape[1]) # Gives the No of columns
```

```python
# Display the information in the dataset
df.info()
```

```python
# Identify numerical and categorical columns
numerical_columns = df.select_dtypes(include=np.number).columns.tolist()
categorical_columns = df.select_dtypes(include='object').columns.tolist()

print("Numerical Columns:")
print(numerical_columns)

print("\nCategorical Columns:")
print(categorical_columns)
```

```python
#summary of numerical columns
print("Statistical Summary:")
display(df.describe())
```

```python
# Analyze categorical features
print("Total Unique Commodities:", df['commodity'].nunique())
print("Total Unique States:", df['state'].nunique())
print("Total Unique Districts:", df['district'].nunique())
print("Total Unique Markets:", df['mkt_id'].nunique())
```

```python
# Convert date column into datetime format
df['date'] = pd.to_datetime(df['date'])
# Find minimum and maximum dates
print("Starting Date:", df['date'].min())
print("Ending Date:", df['date'].max())
```

```python
# Top 10 frequently occurring commodities
top_commodities = df['commodity'].value_counts().head(10)
print("Top 10 Commodities:")
display(top_commodities)
```

```python
# Count market observations state-wise
state_counts = df['state'].value_counts().head(10)
print("Top 10 States by Number of Records:")
display(state_counts)
```

```python
# Count market observations
market_counts = df['market'].value_counts().head(10)
print("Top 10 Markets by Number of Records:")
display(market_counts)
```
