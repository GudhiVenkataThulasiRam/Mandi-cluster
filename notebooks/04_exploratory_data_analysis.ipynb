```python

# Import the required  libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load preprocessed dataset
df = pd.read_csv("../dataset/mandi_cleaned.csv")
print("Dataset Loaded Successfully!")
print("Dataset Shape:", df.shape)
print("\nFirst 5 Records:")
display(df.head())
```

```python
# Analyze commodity distribution
commodity_counts = df['commodity'].value_counts().head(10)
print("Top 10 Commodities by Number of Records:")
display(commodity_counts)
# Visualize commodity distribution
plt.figure(figsize=(12, 6))
sns.barplot(
    x=commodity_counts.values,
    y=commodity_counts.index
)
plt.title("Top 10 Commodities by Number of Records")
plt.xlabel("Number of Records")
plt.ylabel("Commodity")
plt.tight_layout()
plt.show()
```

```python
# Analyze state-wise distribution
state_counts = df['state'].value_counts().head(10)
print("Top 10 States by Number of Records:")
display(state_counts)
# Visualize state-wise distribution
plt.figure(figsize=(12, 6))
sns.barplot(
    x=state_counts.values,
    y=state_counts.index
)
plt.title("Top 10 States by Number of Records")
plt.xlabel("Number of Records")
plt.ylabel("State")
plt.tight_layout()
plt.show()
```

```python
# Analyze market-wise distribution
market_counts = df['market'].value_counts().head(10)
print("Top 10 Markets by Number of Records:")
display(market_counts)
# Visualize market distribution
plt.figure(figsize=(12, 6))
sns.barplot(
    x=market_counts.values,
    y=market_counts.index
)
plt.title("Top 10 Markets by Number of Records")
plt.xlabel("Number of Records")
plt.ylabel("Market")
plt.tight_layout()
plt.show()
```

```python
# Analyze modal price distribution
print("Modal Price Statistics:")
display(df['modal_price_rs_qtl'].describe())
# Visualize modal price distribution
plt.figure(figsize=(12, 6))
sns.histplot(
    df['modal_price_rs_qtl'],
    bins=50,
    kde=True
)
plt.title("Distribution of Modal Prices")
plt.xlabel("Modal Price (Rs per Quintal)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
# Analyze arrival quantity distribution
print("Arrival Quantity Statistics:")
display(df['arrival_t'].describe())
# Visualize arrival quantity distribution
plt.figure(figsize=(12, 6))
sns.histplot(
    df['arrival_t'],
    bins=50,
    kde=True
)
plt.title("Distribution of Arrival Quantity")
plt.xlabel("Arrival Quantity (Tonnes)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
# Analyze average modal price commodity-wise
commodity_avg_price = (
    df.groupby('commodity')['modal_price_rs_qtl']
    .mean()
    .sort_values(ascending=False)
)
print("Top 10 Commodities by Average Modal Price:")
display(commodity_avg_price.head(10))
# Visualize top 10 commodities
plt.figure(figsize=(12, 6))
sns.barplot(
    x=commodity_avg_price.head(10).values,
    y=commodity_avg_price.head(10).index
)
plt.title("Top 10 Commodities by Average Modal Price")
plt.xlabel("Average Modal Price (Rs per Quintal)")
plt.ylabel("Commodity")
plt.tight_layout()
plt.show()
```

```python
# Analyze average modal price state-wise
state_avg_price = (
    df.groupby('state')['modal_price_rs_qtl']
    .mean()
    .sort_values(ascending=False)
)
print("Top 10 States by Average Modal Price:")
display(state_avg_price.head(10))
# Visualize state-wise average prices
plt.figure(figsize=(12, 6))
sns.barplot(
    x=state_avg_price.head(10).values,
    y=state_avg_price.head(10).index
)
plt.title("Top 10 States by Average Modal Price")
plt.xlabel("Average Modal Price (Rs per Quintal)")
plt.ylabel("State")
plt.tight_layout()
plt.show()
```

```python
# Analyze average modal price market-wise
market_avg_price = (
    df.groupby('market')['modal_price_rs_qtl']
    .mean()
    .sort_values(ascending=False)
)
print("Top 10 Markets by Average Modal Price:")
display(market_avg_price.head(10))
# Visualize top 10 markets
plt.figure(figsize=(12, 6))
sns.barplot(
    x=market_avg_price.head(10).values,
    y=market_avg_price.head(10).index
)
plt.title("Top 10 Markets by Average Modal Price")
plt.xlabel("Average Modal Price (Rs per Quintal)")
plt.ylabel("Market")
plt.tight_layout()
plt.show()
```

```python
# Analyze monthly average modal price trends
monthly_df = df.copy()
# Convert date column into datetime format
monthly_df['date'] = pd.to_datetime(
    monthly_df['date'], errors='coerce'
)
# Extract month and year
monthly_df['month'] = (
    monthly_df['date'].dt.to_period('M').astype(str)
)
# Calculate monthly average modal price
monthly_price = (
    monthly_df.groupby('month')['modal_price_rs_qtl']
    .mean()
    .reset_index()
)
print("Monthly Average Modal Price:")
display(monthly_price)
# Visualize monthly price trends
plt.figure(figsize=(14, 6))
sns.lineplot(
    data=monthly_price,
    x='month',
    y='modal_price_rs_qtl',
    marker='o'
)
plt.title("Monthly Average Modal Price Trend")
plt.xlabel("Month")
plt.ylabel("Average Modal Price (Rs per Quintal)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

# Analyze monthly price trends for top 5 commodities
top_commodities = (
    df['commodity'].value_counts().head(5).index
)
commodity_monthly = df[
    df['commodity'].isin(top_commodities)
].copy()
commodity_monthly['date'] = pd.to_datetime(
    commodity_monthly['date'],
    errors='coerce'
)
commodity_monthly['month'] = (
    commodity_monthly['date'].dt.to_period('M').astype(str)
)
monthly_commodity_price = (
    commodity_monthly.groupby(
        ['month', 'commodity']
    )['modal_price_rs_qtl']
    .mean()
    .reset_index()
)
print("Monthly Commodity-wise Average Prices:")
display(monthly_commodity_price.head(15))
# Visualize commodity-wise monthly price trends
plt.figure(figsize=(14, 7))
sns.lineplot(
    data=monthly_commodity_price,
    x='month',
    y='modal_price_rs_qtl',
    hue='commodity',
    marker='o'
)
plt.title("Monthly Price Trends of Top 5 Commodities")
plt.xlabel("Month")
plt.ylabel("Average Modal Price (Rs per Quintal)")
plt.xticks(rotation=45)
plt.legend(title="Commodity")
plt.tight_layout()
plt.show()
```

```python

# Analyze commodity-wise price volatility

commodity_volatility = (
    df.groupby('commodity')['modal_price_rs_qtl']
    .agg(['mean', 'std'])
    .sort_values(by='std', ascending=False)
)

print("Top 10 Commodities by Price Volatility:")
display(commodity_volatility.head(10))

# Visualize price volatility
plt.figure(figsize=(12, 6))

sns.barplot(
    x=commodity_volatility.head(10)['std'].values,
    y=commodity_volatility.head(10).index
)

plt.title("Top 10 Commodities by Price Volatility")
plt.xlabel("Standard Deviation of Modal Price")
plt.ylabel("Commodity")

plt.tight_layout()
plt.show()
```

```python
# Analyze price distribution of top 5 commodities
top_5_commodities = (
    df['commodity'].value_counts().head(5).index
)
commodity_price_data = df[
    df['commodity'].isin(top_5_commodities)
]
# Visualize price distribution
plt.figure(figsize=(12, 6))
sns.boxplot(
    data=commodity_price_data,
    x='commodity',
    y='modal_price_rs_qtl',
    showfliers=False
)
plt.title("Price Distribution of Top 5 Commodities")
plt.xlabel("Commodity")
plt.ylabel("Modal Price (Rs per Quintal)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python

# State-wise commodity price comparison
# Select top 10 states by number of records
top_states = df['state'].value_counts().head(10).index
# Select top 5 commodities by number of records
top_commodities = df['commodity'].value_counts().head(5).index
# Filter the dataset
heatmap_data = df[
    (df['state'].isin(top_states)) &
    (df['commodity'].isin(top_commodities))
]
# Create state-wise and commodity-wise average prices
price_pivot = heatmap_data.pivot_table(
    index='state',
    columns='commodity',
    values='modal_price_rs_qtl',
    aggfunc='mean'
)
print("State-wise Commodity Average Price:")
display(price_pivot)
# Create heatmap
plt.figure(figsize=(14, 8))
sns.heatmap(
    price_pivot,
    annot=True,
    fmt='.0f',
    cmap='YlOrRd',
    linewidths=0.5
)
plt.title("State-wise Commodity Price Comparison")
plt.xlabel("Commodity")
plt.ylabel("State")
plt.tight_layout()
plt.show()
```

```python
# Analyze relationship between arrival quantity and modal price
print("Correlation between Arrival Quantity and Modal Price:")
correlation = df[
    ['arrival_t', 'modal_price_rs_qtl']
].corr()
display(correlation)
# Take a sample to make the scatter plot readable
sample_df = df.sample(
    n=min(5000, len(df)),
    random_state=42
)
# Visualize relationship
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=sample_df,
    x='arrival_t',
    y='modal_price_rs_qtl',
    alpha=0.4
)
plt.title("Relationship Between Arrival Quantity and Modal Price")
plt.xlabel("Arrival Quantity (Tonnes)")
plt.ylabel("Modal Price (Rs per Quintal)")
plt.tight_layout()
plt.show()
```

```python
# Correlation Analysis
import seaborn as sns
import matplotlib.pyplot as plt
# Select numerical columns
correlation_data = df[['arrival_t', 'modal_price_rs_qtl']]
# Calculate correlation
correlation_matrix = correlation_data.corr()
print("Correlation Matrix:")
display(correlation_matrix)
# Visualize correlation using heatmap
plt.figure(figsize=(8, 5))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)
plt.title("Correlation Between Arrival Quantity and Modal Price")
plt.show()
```

```python
# Top 10 Markets by Average Modal Price
market_price = df.groupby(
    ['state', 'district', 'market']
)['modal_price_rs_qtl'].mean().reset_index()
# Sort markets by average price
top_markets = market_price.sort_values(
    by='modal_price_rs_qtl',
    ascending=False
).head(10)
print("Top 10 Markets by Average Modal Price:")
display(top_markets)
# Visualize results
plt.figure(figsize=(12, 6))
sns.barplot(
    data=top_markets,
    x='modal_price_rs_qtl',
    y='market',
    hue='state',
    dodge=False
)
plt.title("Top 10 Markets by Average Modal Price")
plt.xlabel("Average Modal Price (Rs per Quintal)")
plt.ylabel("Market")
plt.tight_layout()
plt.show()
```

```python
# Commodity-wise Price Comparison Across Markets
commodity_market_price = df.groupby(
    ['commodity', 'market']
)['modal_price_rs_qtl'].mean().reset_index()
# Select the top 10 commodity-market combinations
top_combinations = commodity_market_price.sort_values(
    by='modal_price_rs_qtl',
    ascending=False
).head(10)
print("Top 10 Commodity-Market Combinations by Average Price:")
display(top_combinations)
# Visualize results
plt.figure(figsize=(13, 7))
sns.barplot(
    data=top_combinations,
    x='modal_price_rs_qtl',
    y='commodity',
    hue='market'
)
plt.title("Top Commodity-Market Price Comparisons")
plt.xlabel("Average Modal Price (Rs per Quintal)")
plt.ylabel("Commodity")
plt.tight_layout()
plt.show()
```

```python
# Monthly Arrival Quantity Trend
# Convert date column to datetime
df['date'] = pd.to_datetime(df['date'])
# Extract month and calculate total arrivals
df['month'] = df['date'].dt.to_period('M').astype(str)
monthly_arrivals = df.groupby('month')['arrival_t'].sum().reset_index()
print("Monthly Arrival Quantity:")
display(monthly_arrivals)
# Visualize monthly arrivals
plt.figure(figsize=(13, 6))
plt.plot(
    monthly_arrivals['month'],
    monthly_arrivals['arrival_t'],
    marker='o'
)
plt.title("Monthly Agricultural Arrival Quantity Trend")
plt.xlabel("Month")
plt.ylabel("Total Arrival Quantity (Tonnes)")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Monthly Arrival Trends of Top 5 Commodities
# Identify top 5 commodities
top5_commodities = df['commodity'].value_counts().head(5).index
# Filter records
commodity_arrivals = df[
    df['commodity'].isin(top5_commodities)
]
# Calculate monthly arrivals
monthly_commodity_arrivals = commodity_arrivals.groupby(
    ['month', 'commodity']
)['arrival_t'].sum().reset_index()
print("Monthly Arrival Quantity of Top 5 Commodities:")
display(monthly_commodity_arrivals.head(15))
# Create pivot table
pivot_arrivals = monthly_commodity_arrivals.pivot(
    index='month',
    columns='commodity',
    values='arrival_t'
)
# Plot trends
plt.figure(figsize=(14, 7))
for commodity in pivot_arrivals.columns:
    plt.plot(
        pivot_arrivals.index,
        pivot_arrivals[commodity],
        marker='o',
        label=commodity
    )
plt.title("Monthly Arrival Trends of Top 5 Commodities")
plt.xlabel("Month")
plt.ylabel("Total Arrival Quantity (Tonnes)")
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
