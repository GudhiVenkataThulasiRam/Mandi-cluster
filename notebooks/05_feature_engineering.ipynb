```python
# Step 1: Import libraries and load the cleaned dataset
import pandas as pd
import numpy as np
df = pd.read_csv("../dataset/mandi_cleaned.csv")
# Convert date column into datetime format
df['date'] = pd.to_datetime(df['date'])
print("Dataset Loaded Successfully!")
print("Dataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
print("\nFirst 5 Records:")
display(df.head())
```

```python
# Step 2: Aggregate data by market and commodity
market_features = df.groupby(
    ['state', 'district', 'market', 'commodity']
).agg(
    avg_price=('modal_price_rs_qtl', 'mean'),
    median_price=('modal_price_rs_qtl', 'median'),
    std_price=('modal_price_rs_qtl', 'std'),
    min_price=('modal_price_rs_qtl', 'min'),
    max_price=('modal_price_rs_qtl', 'max'),
    avg_arrival=('arrival_t', 'mean'),
    median_arrival=('arrival_t', 'median'),
    total_arrival=('arrival_t', 'sum'),
    total_records=('modal_price_rs_qtl', 'count')
).reset_index()
# Replace missing standard deviation values with 0
market_features['std_price'] = market_features['std_price'].fillna(0)
print("Market-Commodity Features Created Successfully!")
print("Feature Dataset Shape:", market_features.shape)
print("\nFirst 5 Records:")
display(market_features.head())
```

```python
# Step 3: Select numerical features
clustering_features = [
    'avg_price',
    'median_price',
    'std_price',
    'min_price',
    'max_price',
    'avg_arrival',
    'median_arrival',
    'total_arrival',
    'total_records'
]
X = market_features[clustering_features].copy()
print("Numerical Features Selected Successfully!")
print("Feature Matrix Shape:", X.shape)
print("\nSelected Features:")
print(clustering_features)
print("\nMissing Values:")
print(X.isnull().sum())
print("\nFirst 5 Records:")
display(X.head())
```

```python
# Step 4: Check missing and infinite values
print("Missing Values:")
print(X.isnull().sum())
print("\nTotal Missing Values:", X.isnull().sum().sum())
print("\nInfinite Values:")
print(np.isinf(X).sum())
print("\nTotal Infinite Values:", np.isinf(X).sum().sum())
```

```python
# Step 5: Check skewness of numerical features / feature skewness

skewness = X.skew().sort_values(ascending=False)
print("Feature Skewness:")
display(skewness)
print("\nHighly Skewed Features (absolute skewness > 1):")
display(skewness[abs(skewness) > 1])
```

```python
# Step 6: Apply log transformation to highly skewed features
skewed_features = [
    'avg_price',
    'median_price',
    'std_price',
    'min_price',
    'max_price',
    'avg_arrival',
    'median_arrival',
    'total_arrival'
]
X_transformed = X.copy()
for feature in skewed_features:
    X_transformed[feature] = np.log1p(X_transformed[feature])

print("Log Transformation Completed!")
print("\nSkewness After Transformation:")
display(X_transformed.skew().sort_values(ascending=False))
print("\nTransformed Dataset:")
display(X_transformed.head())
```

```python
# Step 7: Combine identifiers and transformed features
identifiers = market_features[
    ['state', 'district', 'market', 'commodity']
].reset_index(drop=True)
engineered_dataset = pd.concat(
    [identifiers, X_transformed.reset_index(drop=True)],
    axis=1
)

# Save the engineered dataset
engineered_dataset.to_csv(
    "../dataset/market_features_engineered.csv",
    index=False
)
print("Engineered Dataset Saved Successfully!")
print("Dataset Shape:", engineered_dataset.shape)
print("\nSaved File: market_features_engineered.csv")
print("\nFirst 5 Records:")
display(engineered_dataset.head())
```
